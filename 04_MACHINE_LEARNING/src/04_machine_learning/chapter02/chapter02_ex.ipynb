```python
import pandas as pd 
import matplotlib.pyplot as plt
import platform
from matplotlib import rc

os_name = platform.system()

if os_name == 'Windows':
    rc("font", family="Malgun Gothic")
else:
    rc("font", family="AppleGothic")

df = pd.read_csv("usa_housing.csv")
df_after_2000 = df[(df['YearBuilt'] > 2000)]
scatter = plt.scatter(x=df_after_2000['Price'], y=df_after_2000['SchoolRating'], c=df_after_2000['SchoolRating'])
plt.xlabel("Price")
plt.ylabel("SchoolRating")
plt.title("학군 점수에 따른 주택 가격 분포")

plt.show()
```

```python
from sklearn.linear_model import LinearRegression

X = df_after_2000[['SchoolRating']]
y = df_after_2000['Price']

model = LinearRegression()
model.fit(X, y)

print(model.coef_, model.intercept_)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

y_pred = model.predict(X=X)
y_pred

mse = mean_squared_error(y, y_pred=y_pred)
mae = mean_absolute_error(y, y_pred=y_pred)

print("MSE:", mse)
print("MSA:", mae)

plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='dodgerblue', alpha=0.5, edgecolor='w', label='Actual Data')
plt.plot(X, y_pred, color='crimson', linewidth=3, label='Regression Line')
plt.title('Linear Regression Prediction')
plt.xlabel('Price')
plt.ylabel('SchoolRating')
plt.legend()
plt.grid(True, linestyle='--')
plt.show()
```
