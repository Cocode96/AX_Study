매개변수 기본값 (Default Parameter)

```javascript
function add(num1, num2 = 10)
{
    return num1 + num2;
}

add(20)
```

arguments 객체

```javascript
function add(...arg)
{
    let result = 0;
    arg.forEach(element => {
        result += element;
    });

    return result;
}

add(10,20,30,50,40,60,80);
```

나머지 매개변수 (Rest Parameter, 가변 매개변수)

```javascript
const nums1 = [10,20,30,40];
const nums2 = [...nums1];

console.log(nums2);
```

```javascript
const person = {name: "김철수", age:20};
const person2 = {...person};

console.log(person2, person === person2)
```

```javascript
const person3 = {...person, address : "주소"};

console.log(person3)
```

```javascript
const [a,,c] = nums1;
console.log(a, c);
```

```javascript
const [a,b, ...rest] = nums1;
console.log(a,b, rest);
```

```javascript
const person = {name: "김철수", age:20};
const name = person.name;
const age = person.age;

console.log(name, age);
```

```javascript
const {name : src, age : tmp} = person;

console.log(src, tmp);
```

#### 4. 값으로서의 함수 (일급 객체)와 콜백

#### 4.1 용어의 유래: 차별 없는 '동등한 권리’

#### 4.2 일급 객체로서의 3대 속성
- 변수 할당 (Assignment)
- 매개변수 전달 (Argument Passing)
- 반환값 활용 (Return Value)

```javascript
function func1(){
    console.log("func1()");
}

function func2(f){
    console.log("func2()");
    f();
}

func2(func1);
```

```javascript
function square(num){
    return num * num;
}

function calc(nums, f){
    if(typeof f === "function")
    {
        throw new TypeError("f는 함수여야합니다.")
    }

    let total = 0;
    for (const num of nums)
        total += f(num);
    
    return total;
}

calc([10,20,30], square);
```

```javascript
calc([10, 20, 30], function(num){
    return num*num;
})
```

```javascript
calc([10,20,30], (num) => {
    return num*num;
})
```

```javascript
function f(a) {
    console.log(a);
    console.log(arguments[0]); // 10
    console.log(arguments[1]); // 20
    console.log(arguments[2]); // 30
}

f(10, 20, 30);
```

### 5. 화살표 함수 (Arrow Function)

#### 5.1 정의 방법 및 스코프 생략 규칙

```javascript
calc([10,20,30], (num) => {
    return num * num;
})
```

#### 5.2 화살표 함수의 제한 사항
- 생성자 함수 사용 불가
- arguments 객체 미지원
- yield 사용 불가

```javascript
const obj = {
    name: "Cloud",

    arrow: () => {
        console.log(this.name);
    }
};

obj.arrow();
```

#### 5.3 일반 함수 vs 화살표 함수의 this 바인딩 비교

```javascript
const person = {
    showinfo() {
        const f = () => {
            console.log(this);
        };

        f();
    }
};

person.showinfo();
```
