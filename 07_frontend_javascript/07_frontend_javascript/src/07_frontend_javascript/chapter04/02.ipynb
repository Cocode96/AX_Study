### 2. 초기 자바스크립트의 변수 선언: var의 치명적인 결함

#### 2.1 var의 3가지 결함
- 호이스팅 (Hoisting)
- 선언 키워드 생략 허용 (암묵적 전역 변수)
- 함수 레벨 스코프 (Function-level Scope)

```javascript
if(true)
{
    var hp_var = 100;
    let hp_let = 100;
}

console.log(hp_var); // 함수
// console.log(hp_let); 이건 오류남 let은 스코프에서만 유효하다.
```

```javascript
for (var i = 0; i < 3; i++) {
    setTimeout(() => console.log(i), 100);
}

// var : 하나의 i 바인딩을 모든 콜백이 공유해서 최종값 3을 읽음
// 3
// 3
// 3 
```

### 3. 현대 자바스크립트의 해결책: let과 const (ES6+)

#### 3.1 let과 const의 핵심 특징
- 블록 레벨 스코프 (Block-level Scope)
- 호이스팅 사전 차단

```javascript
for (let i = 0; i < 3; i++) {
    setTimeout(() => console.log(i), 100);
}

// let : 반복마다 새로운 i 바인딩이 생성되고 각 콜백이 해당 i를 캡쳐해서 0, 1, 2
// 0
// 1
// 2
```

### 4. 데이터 타입과 typeof를 활용한 타입 검증

#### 4.1 자바스크립트의 주요 데이터 타입
- Number (숫자형)
- String (문자열)
- Boolean (불리언)
- null
- undefined

```javascript
const num1 = 100;
console.log(typeof num1); // "number"

const str1 = "ABC";
console.log(typeof str1); // "string"

const bool1 = true;
console.log(typeof bool1); // "boolean"

const null1 = null;
console.log(typeof null1); // "object"

const undefined1 = undefined;
console.log(typeof undefined1); // "undefined"
```

```javascript
const a = {};
const b = {};
const c = a;

console.log(a === b); // false
console.log(a === c); // true

console.log(a == b); // false
console.log(a == c); // true
```
