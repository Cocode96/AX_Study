```python
import time

def download_page(data_name, delay_time):
    print("시작")
    time.sleep(delay_time)
    print("완료")

download_page("메인 페이지", 2)
download_page("상세 페이지", 2)
```

```python
import asyncio, time

async def download_page_async(data_name, delay_time):
    print("시작")
    await asyncio.sleep(delay_time)
    print("완료")

async def main():
    start = time.perf_counter()

    await asyncio.gather(
        download_page_async("메인 페이지", 2),
        download_page_async("상세 페이지", 2)
    )

    end = time.perf_counter() - start
    print("걸린 시간 : ", end)

await main()
```

```python
async def fetch_weather(local_name, delay_time):
    print(f"{local_name} 수집 시작")
    await asyncio.sleep(delay_time)
    print(f"{local_name} 수집 완료")

async def main():
    start = time.perf_counter()

    await asyncio.gather(
        fetch_weather("서울", 1.5),
        fetch_weather("뉴욕", 2.5),
        fetch_weather("런던", 1)
    )

    end = time.perf_counter() - start
    print("걸린 시간 : ", end)

await main()
```

```python
async def cook_ramen():
    print("라면 조리를 시작합니다.")
    await asyncio.sleep(3)
    print("라면 조리가 완료되었습니다.")

await cook_ramen()
```

```python
async def download_music(music_name, wait_seconds):
    print("다운로드를 시작합니다.")
    asyncio.sleep(wait_seconds)
    print("다운로드가 완료되었습니다.")

async def main():
    asyncio.gather(
        download_music("첫번째 음악", 2),
        download_music("두번째 음악", 4)
    )

asyncio.run(main())
```
