### 1. 항과 연산자
- 항 : 연산에 사용되는 값, 2개 -> 이행, 3개 -> 삼항, 1개 -> 단항
- 연산자 : 연산에 사용되는 기호(+, =, ...)

### 2. 연산자의 종류

#### 대입 연산자
x = 10 대입연산자는 우선순위가 낮음

#### 산술 연산자
+, -, *, /, // : 나눈 몫, %, ** : 거듭제곱

- 연산 우선순위 : **, *, /, //, % > +, -
- 우선 순위 강제 적용 : (...)

```python
num1 = 10
num2 = 3

num1 // num2, num1 % num2
```

```python
num1 = 3

num1 ** 3
```

#### 비교 연산자
\>, <, >=, <=
== : 같다
!= : 다르다

- 연산의 결과는 무조건 True or False
- 판별이 필요한 구문(조건문, 반복문)  

```python
10 > 3
```

```python
num1 = 1000
num2 = 1000
hex(id(num1)), hex(id(num2))
```

```python
num1 == num2 # 동등성 비교
```

```python
num1 is num2 # 동일성 비교(주소 비교)
```

```python
num1 = 20

result1 = num1 > 10
result2 = num2 < 100

result1 and result2
```

```python
not (result1 and result2)
```

#### 논리 연산자(and, or, not)
and
- 각 항이 모두 true가 되어야지 true가 나온다

or
- 둘중하나가 

```python
result1 or result2
```

```python
fruit1 = "사과"
fruit2 = "오렌지"
```

```python
fruit1 == "사과" or fruit2 == "오렌지"
```

#### 복합 대입 연산자
대입연산자(=) + 다른 연산자(주로 산술 연산자 -, +, *, /, //, %, **)결합

```python
num = 1
num = num + 1

num
```

```python
 num = 1
 num += 1

 num
```

### 3. 단락 회로 평가(Short-Circuit Evaluation)

```python
result = True or 10**2 # 첫항의 결과가 True-> 결과가 이미 결정 -> 다음 연산 진행 x

result
```

```python
result = False or 10**2 # 첫항의 결과가 False -> 결과 아직 결정 x -> 다음 항 체크(연산)

result
```

```python
# or 연산을 이용한 기본값(없을때 대체되는 값)
message = ""

result = message or "기본 메세지"
result
```

```python
# and 연산을 이용한 간단한 조건 처리
height = 165

result = height >= 160 and "크다"
result
```
