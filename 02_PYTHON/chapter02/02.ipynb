### 1. 파이썬 기본 자료형 (Primitive Data Types)

- 자료 : 연산에 필요한 값(데이터)
- 자료형 : 값의 유형(테이터의 종류)

#### 자료형
- 숫자형 : 정수형(int, 10, 20, ...), 실수형(float, 10.123, ...)
- 논리형 : 참(True), 거짓(False)
- 문자형 : 문자, 문장, '', ""
- 값 없음 : None

```python
moeny = None
```

```python
message = """
안녕하세요.
반갑습니다.
3줄이에요.
"""
type(message)
```

### 2. 명시적 형변환(Explicit Type Conversion)

연산은 같은 자료형끼리 가능

형변환 함수
-int(...) : 정수로 변환
-float(...) : 실수로 변환
-bool(...) : True, False로 변환
-str(...) : 문자열로 변환

```python
num = "100"
type(num)
```

```python
int(num) + 200
```

```python
"현재 시간은 " + str(10) + "시"
```

### 3. 묵시적 형변환(Implicit Type Conversion)

표현 범위가 더 넓은 자료형으로 형변환된다.

```python
num1 = 100
num2 = 3.123

type(num1), type(num2)
```

```python
num3 = num1 + num2 # num1이 int -> float 자동 형변환(묵시적 형변환)
#num3 = num1 +int(num2)

num3, type(num3)
```

### 4. 형변환 시 주의해야 할 예외와 에러(ValueError)

```python
time = "10시"
if time.isdigit() == True:
    int(time) + 1
```

### 5. 거짓으로 인식하는 값
0, "", [], None

```python
message = ""
if not message:
    print("메세지 없음")

if message == False:
    print("이건 안나옴")

if message == "":
    print("보이는데로 message는 "" 이기때문에 나온다.")
```

```python
result = bool(message)

result, type(result)
```
