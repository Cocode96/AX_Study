#### 1. 1세대 : % 연산자 방식

- %s : 문자열 대체
    - %숫자s : 최소 숫자 자리수 만큼 공간을 확보, 남는 공간은 공백
- %d : 정수
    - %대체문자숫자d -> %05d -> 남는 공백은 0을 대체
- %f : 실수
    - %.2f : 소수점 2자리까지 표시

```python
message = "나의 이름은 %5s이고, 나이는 %d입니다"

message % ("철수", 20)
```

```python
message = "주문번호 : %05d" % 10
message
```

```python
#message = "숫자 : %f" % 10.123154
#message = "숫자 : %.2f" % 10.123154
message = "숫자 : %010.2f" % 10.123154

message
```

```python
message = "숫자 : %d" % 10.12345

message
```

```python
message = "숫자 : %s" % 10.12345
message
```

#### 2. 2세대 : str.format() 메서드

- 순서 지정 교체 : {0} {1} {2}

```python
message = "내 이름은 {0}이고, 나이는 {1}입니다."

message.format("철수", 20)
```

```python
message = "내 이름은 {name}이고, 나이는 {age}입니다."

message.format(name ="철수", age="20")
```

#### 3. 3세대 : f-string 방식

```python
name = "영희"
age = 20

message = f"내이름은 {name}이고, 나이는 {age}입니다."

message
```

```python
price = 10000

f"가격 : {price:,}"
```

```python
f"가격 : {price:>10,}원"
```

```python
#디버깅 조절자(=)
f"{price=}"
```
