```python
menu_name = "초코라떼"
menu_price = 12000
quantity = 2

# 순서대로 인덱스 번호를 지정하여 대입
indexed_format = "주문 내역 : {0} {1}잔 -> 총 {2}원".format(menu_name, quantity, menu_price * quantity)
print(indexed_format)

# 중괄호 안에 이름표(key)를 지정하여 대입 (가독성 향상)
# {item:>12}: 12칸을 채운 뒤 오른쪽 정렬
named_format = "메뉴 : {item:>6} | 결제금액 : {price}원".format(item=menu_name, price=menu_price * quantity)
print(named_format)
```

```python
name = "Cheese cake"
price = 5000
qty = 1

# 문자열 안에 변수와 가벼운 연산식을 직접 삽입
print(f"주문하신 {name} {qty}개의 최종 청구액은 {price}입니다.")

# 3자리 단위 콤마(,) 및 오른쪽 정렬 제어
# :>10, -> 10칸의 자리를 확보하고 오른쪽 정렬한 후 천 단위 쉼표 추가
print(f"포맷팅 적용 : {name:>10} | 가격 : {price:>4,}")

# 디버깅 전용 조절자 (=) 사용
# 변수 이름과 현재 값을 동시에 출력하여 로그 확인 시 유용
current_status = "정상 진행중"
print(f"디버깅 로그 -> {current_status=}, {price=}")

# 최신 Python 3.12/3.13 기능 (PEP 701)
# 바깥쪽과 안쪽 모두 큰따옴표("")를 중복 사용해도 에러가 나지 않음
menu_dict={"name" : "Cheese cake", "price" : 5000, qty : 1}
print(f"딕셔너리 직접 출력 : {menu_dict["name"]}의 가격은 {menu_dict["price"]}")
```

```python
# 가상의 주문 데이터 리스트
orders = [
    ("Cafe Latte", 5000, 2),
    ("Americano", 4500, 1),
    ("Cheese cake", 6500, 1)
]

print("=" * 35)
# '메뉴명'은 왼쪽 정렬(12칸), '단가', '수량', '금액'은 오른쪽 정렬합니다.
print(f"{'메뉴명':<12}{'단가':>8}{'수량':>4}{'금액':>9}")
print("-" * 35)

total_receipt_amount = 0
for menu, item_price, qty in orders:
    subtotal = item_price * qty
    total_receipt_amount += subtotal

    print(f"{menu:<12}{item_price:>8}{qty:>4}{subtotal:>9}")


print("-" * 35)
print(f"{'합계 금액':<12}{total_receipt_amount:>21,}원")
print("=" * 35)
```
