```python
winner_names = ["이민서", "박준형"]
winner_email = ["minseo@ssu.ac.kr", "junhyung@daum.net"]

zipped = zip(winner_names, winner_email)
enumerated = enumerate(zipped, start=1)
for index, (name, email) in enumerated:
    print(f"{index}번 당첨자 : {name} | 이메일 : {email}")
```
