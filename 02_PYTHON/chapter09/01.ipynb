### 2. all()과 any()
- all(iterable) : 모든 요소가 True로 인식될때 -> True 예외) 빈 리스트([]) - True
- any(iterable) : 요소중 하나라도 True 인식이 되면 -> True

```python
all([True, 1, " "])
```

```python
all([])
```

```python
any([False, 1, 0])
```

```python
any([])
```

### 3. len(), sum(), round()
- len() : 문자열 -> 문자열의 길이, 리스트, 튜플 -> 요소의 갯수
- sum(iterable) : 합계(요소는 숫자로만 구성)
- round() : 반올림, Math : ceil(...) : 올림, floor(...) 버림

```python
num = 10.56
print(round(num, 1))

import math

print(math.ceil(0.2))
print(math.floor(num))
```

### 4. mean()과 median()
- statistics 모듈에 포함된 표준 라이브러리
- mean(iterable) : 평균
- median(iterable) : 중앙값

```python
import statistics
nums = [10, 15, 20, 8, 7, 10000000]
print("평균 : ", statistics.mean(nums), "중앙값 : ", statistics.median(nums))
```
