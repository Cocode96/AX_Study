### 2. all()과 any()
- all(iterable) : 모든 요소가 True로 인식될때 -> True 예외) 빈 리스트([]) - True
- any(iterable) : 요소중 하나라도 True 인식이 되면 -> True

```python
all([True, 1, " "])
```

```python
all([])
```

```python
any([False, 1, 0])
```

```python
any([])
```

### 3. len(), sum(), round()
- len() : 문자열 -> 문자열의 길이, 리스트, 튜플 -> 요소의 갯수
- sum(iterable) : 합계(요소는 숫자로만 구성)
- round() : 반올림, Math : ceil(...) : 올림, floor(...) 버림

```python
num = 10.56
print(round(num, 1))

import math

print(math.ceil(0.2))
print(math.floor(num))
```

### 4. mean()과 median()
- statistics 모듈에 포함된 표준 라이브러리
- mean(iterable) : 평균
- median(iterable) : 중앙값

```python
import statistics
nums = [10, 15, 20, 8, 7, 10000000]
print("평균 : ", statistics.mean(nums), "중앙값 : ", statistics.median(nums))
```

### 5. map()과 filter()
- map() : 변환 작업
- filter() : 판별을 통해서 정제해주는거

-> 리스트 컴프리헨션으로 대체 가능

```python
nums = [*range(1,11)]
print(nums)
nums2 = [x**2 for x in nums]
print(nums2)
result = list(map(lambda x: x**2, nums))
result2 = list(map(lambda x : x // 2, filter(lambda x: x % 2 != 0, nums)))
print(result)
print(result2)
```

### 6. enumerate()
- 열거체

```python
fruits = ['사과', '오렌지', '멜론', '망고']

num = 1
for fruit in fruits:
    print(num, fruit)
    num += 1
```

```python
for item in enumerate(fruits, start=1):
    print(*item)

print(*enumerate(fruits))
```

### 7. zip()

```python
persons = ["김철수", "김영희", "김민수"]
fruits = ["사과", "오렌지", "멜론"]

print(hex(id(zip(persons, fruits))))
item = zip(persons, fruits)
print(*item)

print(hex(id(item)))
print(hex(id(zip(persons, fruits))))
```

### 8. sorted()
- 리스트.sort() : 원본 데이터를 정렬
- sorted(Iterable) : 원본 데이터는 유지하고, 새로운 객체를 생성해서 정렬(불변 객체)

```python
import random

nums = [*range(1, 11)]

random.shuffle(nums)

print(nums)
nums.sort()
print(nums)

nums_sorted = sorted(nums)
nums_sorted

print(hex(id(nums)))
print(hex(id(nums_sorted)))
```

### 9. min(), max()
- min() : 최소값 / 오름 차순 정렬 후 가장 처음 요소
- max() : 최대값 / 오름 차순 정렬 후 가장 마지막 요소

```python
random.shuffle(nums)
nums
```

```python
print(min(nums), max(nums))
print(nums)
```
