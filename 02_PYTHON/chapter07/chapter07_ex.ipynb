```python
print("태어난 연도를 입력하세요 : ", end="")
try:
    year = int(input())
except ValueError:
    print("숫자만 입력할 수 있습니다.")
else:
    print(f"나이 : {2026 - year}")
```

```python
menu_list = ["햄버거", "피자", "치킨"]
print("인덱스를 입력하세요.")
try:
    index = int(input())
except ValueError:
    print("잘못된 값입니다. 조회 번호는 정수입니다.")
except IndexError:
    print("잘못된 조회번호입니다.")
else:
    print(f"메뉴 : {menu_list[index]}")
finally:
    print("종료")
```

```python
def check_order_count(order_count):
    if order_count < 0:
        raise ValueError("주문 수량 오류")

    print("수량 설정 완료")

try:
    check_order_count(-1)
except ValueError as e:
    print(e)
```

```python
def calculate_share(total_price, num_people):
    share_per_person = total_price / num_people
    return share_per_person

def start_settlement():
    participants = 0
    result = calculate_share(50000, participants)
    print(result)

start_settlement()
```

```python
def calculate_revenue():
    total_revenue = 0
    daily_sales = [12000, 25000, 31000]
    for amount in daily_sales:
        total_revenue += amount
    print(total_revenue)

calculate_revenue()
```

```python
def sum_scores():
    final_score = 50
    bonus_scores = [10, 20, 30]
    for score in bonus_scores:
        final_score += score
    return final_score

result = sum_scores()
```
