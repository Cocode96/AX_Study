#### 1. 구문 오류(Syntax Error)와 런타임 예외(Exception)
구문 오류(Syntax Error)
- 파이썬 문법 사용상의 문제

런타임 예외(Exception)

#### 3. try-except-else-finally
예외처리 구문
```python
try:
    # 예외가 발생할 가능성이 있는 코드
except:
    # 예외가 발생하면 유입되는 구간, 예외에 대한 대안 처리
    #대안 로직, 로깅 처리
else:
    # 예외가 발생하지않아야 유입되는 구간
finally:
    # 예외가 발생하든 정상 처리되든 항상 유입되는 구간, 공통적인 처리
```

````python
try:
    num = int(input())
    result = 10 / num
    print(result)
    #print("로깅처리...")
except:
    print("0이 아닌 정수를 입력하세요.")
    #print("로깅 처리...")
else:
    print("정상처리")
finally:
    print("로깅처리...")
````

````python
try:
    num = int(input())
    result = 10 / num
    print(result)
    #print("로깅처리...")
except ZeroDivisionError:
    print("0이 아닌 정수를 입력하세요.")
    #print("로깅 처리...")
except ValueError:
    print("숫자가 아닙니다.")
else:
    print("정상처리")
finally:
    print("로깅처리...")
````

````python
try:
    num = int(input())
    result = 10 / num
    print(result)
    #print("로깅처리...")
except ZeroDivisionError, ValueError:
    print("0이 아닌 정수를 입력하세요.")
    #print("로깅 처리...")
else:
    print("정상처리")
finally:
    print("로깅처리...")
````

#### 6. raise를 활용한 데이터 유효성 검증
- 예외를 발생 시키기
```python
raise 예외클래스("메세지..")
```

````python
def login(user_id, user_pass):
    if not (user_id == "user01")
        print("아이디가 일치하지 않습니다.")
        return
    
    if not (user_pass == "1234"):
        print("비밀번호가 일치하지 않습니다.")
        return

    # 로그인
    print("로그인 처리 로직...")
````

````python
def login(user_id, user_pass):
    if user_id != "user01":
        raise ValueError("아이디가 일치하지 않습니다.") # 예외가 발생하고, 코드 실행 흐름은 중단

    if user_pass != "1234":
        raise ValueError("비밀번호가 일치하지 않습니다.")
    
    print("로그인 처리 로직...")
````

````python
try:
    login("user02", 1234)
except ValueError as e:
    print(e)
    print("회원 페이지에서 추가적으로 할 로직")
    
````

#### 7. 사용자 정의 예외
- Exception은 모든 예외의 부모클래스
- Exception을 상속 받으면 또는 Exception을 상속받은 자식 클래스를 상속 받으면 그 클래스도 예외 클래스가 된다.

````python
class UserIdMismatchError(Exception):
    def __str__(self):
        return "아이디가 일치하지 않습니다."

class UserPassMismatchError(Exception):
    def __str__(self):
        return "비밀번호가 일치하지 않습니다."

class Tmp():
    pass

#raise Tmp() # 타입 에러

def login(user_id, user_pass):
    if user_id != "user01":
        raise UserIdMismatchError() # 예외가 발생하고, 코드 실행 흐름은 중단

    if user_pass != "1234":
        raise UserPassMismatchError()
    
    print("로그인 처리 로직...")

try:
    login("user01", 12345)
except (UserIdMismatchError, UserPassMismatchError) as e:
    print(e)
````

#### 8. traceback
- traceback.print_exc() -- 에러 스택 콘솔 출력
- traceback.format_exc() -- 에러 스택을 문자열로 반환
- traceback.print_stack() -- 에러 없이 현재 호출 흐름 확인

````python
import traceback

try:
    result = 10/0
    print(result)
except:
    print("에러 발생")
    #traceback.print_exc()
    message = traceback.format_exc()
    print(message)
````
