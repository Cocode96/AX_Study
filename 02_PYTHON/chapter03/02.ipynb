#### 1. 딕셔너리(Dictionary) 자료구조
- key - value 쌍 구조
- key : value 를 찾기 위한 키워드, 중복 x, 리스트는 사용 불가
- value : key로 찾는 값, 중복 o
- {"키" : 값}

- 추가(CREATE) : 없는 키이름으로 값을 대입하면 추가
- 조회(READ) : ['키 이름'] - 키가 없으면 KeyError가 발생
    - get(키 이름) : 키가 업승면 None 반환
    - get(키 이름, 기본값) 0 키 이름으로 조회가 되지 않으면 설정된 기본값으로 대체
    - keys(키 이름) : 모든 키를 담은 View, 읽기 전용
    - values() : 모든 값을 담은 View, 읽기 전용
    - items() : (키, 값) 목록을 담은 View
- 수정(Update) : 이미 있는 키값에 대입하면 수정
    - update({key, value, key, value})

- 삭제(Delete) : 
    - del item[키]
    - clear() : 전체 비우기

다른 키-값 쌍 자료형 -> 딕셔너리로 변환
dict(...)

#### 3. 집합(Set) 자료구조
- 중복을 허용하지 않는다.
- 순서가 결정되지 않는다.
= {값1, 값2, 값3, ...}

#### 집합 연산
- 합집합 : a | b -> a.union(b)
- 교집합 : a & b -> a.intersection(b)
- 차집합 : a - b -> a.defference(b)
- 대칭차집합 : a ^ b -> a.symmetric_difference(b)

#### 집합 자료로 형변환
- set(리스트 또는 튜플)

```python
a = {10, 10, 10, 20, 30}
a
```

```python
b = {20, 30, 50, 50, 50}
result = a | b
print(result)

result_union = a.union(b)
print(result_union)
```

```python
result2 = a & b
print(result2)

result2_intersection = a.intersection(b)
print(result2)
```

```python
result3 = a - b
print(result3)

result3_diff = a.difference(b)
print(result3_diff)
```

```python
result4 = a ^ b
print(result4)

result4_sym_diff = a.symmetric_difference(b)
print(result4_sym_diff)
```

#### 4. 자료구조 간의 상호 변환과 데이터 정제
- 생성 함수를 통한 변환
    - list(), tuple(), set(), dict() 등의 변환 내장 함수를 활용하면 시퀀스 자료구조와 집합 자료구조의 형태를 자유롭게 변환할 수 있다.

- 중복 제거 정제
    - 대량의 사용자 로그 테이터가 담긴 리스트에서 중복을 완전히 제거하고 청정 고유리스트를 추출 하고싶다면, 리스트 -> 집합(중복 자동 제거) -> 리스트 형태의 파이프라인 작동 방식을 적용

```python
# 실무 데이터 정제 패턴
raw_emails = ["a@sparta.com", "b@sparta.com", "a@sparta.com"]

# 1단계: 집합으로 변환하여 중복 제거
unique_set = set(raw_emails) 

# 2단계: 다시 리스트로 돌려놓아 슬라이싱 및 요소 접근을 위한 순서적 활용 보장
refined_emails = list(unique_set)
print(refined_emails)  # 출력: ['a@sparta.com', 'b@sparta.com']
```
