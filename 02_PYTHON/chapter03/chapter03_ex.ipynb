```python
fruits = ['apple', 'banana', 'cherry']
fruits[1] = "blueberry"
print(*fruits)

colors = ('red', 'green', 'blue')
print(*colors[2:1:-1])
```

```python
group1 = [1, 2]
group2 = [3, 4]

print(*(group1 + group2))

user_roles = ['admin', 'editor', 'member']
print('admin' in user_roles)
```

```python
numbers = [0, 1, 2, 3, 4, 5]
print(numbers[1:4:])

numbers2 = [10, 20, 30, 40]
first, *others = numbers2
print(first, others)
```

```python
raw_emails = ["a@sparta.com", "b@sparta.com", "a@sparta.com"]

refined_emails = list(set(raw_emails))
print(refined_emails)
```

```python
menu = {"coffee" : 4000, "tea" : 4500}

print(menu.keys())
print(menu.values())

set_a = {1, 2, 3}
set_b = {3, 4, 5}

print(set_a & set_b)
print(set_a | set_b)
print(set_a - set_b)
```
