```python
import builtins


def main() -> None:
    namespace: dict[str, object] = {"__builtins__": builtins}

    exec("input = '이미 입력받은 문자열'", namespace)
    print(f"shadowed type={type(namespace['input']).__name__}")

    try:
        exec("player_input = input('입력: ')", namespace)
    except TypeError as error:
        print(f"renaming target does not recover input: {error}")

    print(f"builtins.input is still callable={callable(builtins.input)}")

    exec("del input", namespace)
    exec("restored_input = input", namespace)
    print(f"del restores normal lookup={namespace['restored_input'] is builtins.input}")

    exec("input = '다시 가려진 이름'", namespace)
    namespace["builtins"] = builtins
    exec("explicit_input = builtins.input", namespace)
    print(
        "builtins.input can be accessed explicitly="
        f"{namespace['explicit_input'] is builtins.input}"
    )


if __name__ == "__main__":
    main()
```
