#### 1. 독립 소스 파일 모듈(Module)
- 모듈 - 파이썬 파일 하나
- 다른 파일에서 모듈을 불러와서 사용하는 경우가 많다.
- 모듈명은 파일 이름
- __name__ : 모듈 이름

```python
import mod1
```

#### 2. 디렉터리형 패키지(Package)
- 패키지는 모듈을 모아 놓은 디렉토리
- from 패키명.하위패키지 form ...
- 전체 패키지 경로 : 절대 경로 접근 예) chapter06.mod1
- 상대 경로 접근법(..): 바로 위 패키지
                (.) : 현재 모듈이 잇는 패키지

#### 3. import, from, as 키워드
- import : 모듈 불러오기
- 불러온 모듈에서 마침표(.)를 통해서 접근
- from 모듈명/패키지명 import 모듈에 있는 함수, 변수, 클래스 등등, 다른 모듈
- from 모둘명 import * -> 모듈의 전체 함수, 변수, 클래스를 불러온다. 
- as : alias - 별칭

```python
import mod1 

print(mod1.add(1, 2))
print(mod1.VERSION)
```

```python
from mod1 import VERSION
from mod1 import add
from mod1 import minus

print(VERSION)
print(add(1, 2))
print(minus(2, 3))
```

```python
from mod1 import *

print(VERSION)
print(add(1, 2))
print(minus(2, 3))
```
