```python
def rent_book(user_name, book_title):
    return print(f"{user_name} 회원과 {book_title}책")

rent_book('홍길동', '파이썬 안내서')
rent_book(user_name='이영희', book_title='데이터 구조')
```

```python
def calculate_total(price, delivery_fee=3000):
    total_price = price + delivery_fee
    return f"총 금액 {total_price}", f"결제가 완료되었습니다."

total_price_message, payment_complete_message = calculate_total(25000)
print(total_price_message)
print(payment_complete_message)
```

```python
def get_total_and_average(*arg):
    if len(arg) == 0:
        return 0,0
    return sum(arg), sum(arg) / len(arg)

get_total_and_average(70, 80, 90, 100, 65)
```

```python
total_score = 0

def add_score():
    global total_score
    total_score += 10
    return

add_score()
add_score()

print(total_score)
```

```python
base_numbers = [1,2,3,4,5,6]

multiplied_numbers = list(map(lambda x : x * 3, base_numbers))
print(multiplied_numbers)

final_filtered_numbers = list(filter(lambda x : x > 10, multiplied_numbers))
print(final_filtered_numbers)
```

```python
def message_store():
    storage = ""

    def add_message(message):
        nonlocal storage
        storage += message + '\n'
        return print(storage)

    return add_message

my_storage = message_store()
my_storage("새로운 알림이 도착했습니다.")
my_storage("출석 체크가 완료되었습니다.")
```

```python
def deco_border(callback):

    def wrapper():
        print("=" * 6)
        callback()
        print("=" * 6)

    return wrapper

@deco_border
def show_welcome():
    print("환영합니다 스파르타님")

show_welcome()
```

```python
def auto_tracker(callback):

    def wrapper(*args, **kwargs):
        print("디버그 로그 -> 함수이름: ", __name__)
        print("디버그 로그 -> args : ", *args)
        print("디버그 로그 -> kwargs : ", *kwargs)

        result = callback(*args, *kwargs)

        print("함수가 종료되었습니다.")
        return result

    return wrapper

@auto_tracker
def add_numbers(a, b):
    return a + b

print(add_numbers(1, 2))
```
