```python
def rent_book(user_name, book_title):
    return print(f"{user_name} 회원과 {book_title}책")

rent_book('홍길동', '파이썬 안내서')
rent_book(user_name='이영희', book_title='데이터 구조')
```

```python
def calculate_total(price, delivery_fee=3000):
    total_price = price + delivery_fee
    return f"총 금액 {total_price}", f"결제가 완료되었습니다."

total_price_message, payment_complete_message = calculate_total(25000)
print(total_price_message)
print(payment_complete_message)
```

```python
def get_total_and_average(*arg):
    if len(arg) == 0:
        return 0,0
    return sum(arg), sum(arg) / len(arg)

get_total_and_average(70, 80, 90, 100, 65)
```
