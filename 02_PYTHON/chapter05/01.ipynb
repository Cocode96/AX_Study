#### 2. 매개변수(Parameter)와 인수(Argument)
- 매개변수 : 함수 호출시 입력 부분에서 값을 담아주는 변수
- 인수 : 호출시 사용된 값 (10)

````python
def add(num1, num2):
    num3 = 30
    result = num1 + num2 + num3
    print(hex(id(result)))
    return result

result = add(10, 20)
print(result)
print(hex(id(result)))
````

````python
def func(num):
    result = add(10, 20)
    result += num
    return result

func(5)
````

#### 3. 위치 인수(Positional Arguments)와 키워드 인수(keyword Arguments)
- 위치 인수 : 매개변수의 순서대로 값을 지정
- 키워드 인수 : 

````python
def add(num1, num2):
    return num1 + num2
````

````python
add(10, 20) # 위치 인수
````

````python
add(num1=10, num2=20) # 키워드 인수
````

````python
add(num2=10, num1=20) # 키워드 이름이 중요하지 순서는 상관없음
````

````python
def add(num1, num2, name, mobile):
    return f"{num1 + num2}, {name=}, {mobile=}"
````

````python
add(10, 20, mobile="010-0000-0000", name="김영희")
add(mobile="010-0000-0000", 10, 20,name="김영희")
````

#### 4. 기본값 매개변수(Default Parameter)
매개변수에 값을 할당하지 않아도 기본적으로 가질 수 있는 값 설정
Default Parameter는 오른쪽부터 설정해줘야함

````python
def add(num1, num2, name, mobile = "010"):
    return f"{num1 + num2}, {name=}, {mobile=}"
````

````python
add(10, 20, "김영희")
````

````python
def add(num1, num2, name, mobile = None):
    return f"{num1 + num2}, {name=}, {mobile=}"
````

````python
add(10, 20, "김영희")
````

#### 5. 가변 인수(*args)와 키워드 가변 인수(**kwargs)

````python
def add(*nums):
    print(nums, type(nums))
    total = 0
    for num in nums:
        total += num
    return num

print(add(10))
print(add(10,20))
print(add(10,20,10))
print(add(10,20,10,20))
````

````python
def print_user_info(**info):
    print(info)

print_user_info(name="김철수", age=20)
````

````python
def print_user_info(name, age, **info):
    print("name : ", name)
    print("age : ", age)
    print("info : ", info)

print_user_info("김철수", 20, mobile="010-0000-0000")
````

````python
def print_user_info(name, age, *arg, **kwargs):
    print("name : ", name)
    print("age : ", age)
    print(arg)
    print(kwargs)

print_user_info("김철수", 20, mobile="010-0000-0000")
````

##### 7. 심화 파이썬 함수의 메모리 생에주기(Heap vs Call Stack)

````python
import sys

def outer_func():
    x = 10

    def inner_func():
        y = 20

        # 1. 현재 실행 중인 프레임 객체 가져오기
        frame = sys._getframe()

        print(f"1. 현재 실행 함수: {frame.f_code.co_name}")
        print(f"2. 현재 지역 변수(f_locals): {frame.f_locals}")
        print(f"3. 전역 스코프 참조 여부: {'outer_func' in frame.f_globals}")

        # 2. f_back을 통해 부모 프레임(outer_func)으로 이동
        parent_frame = frame.f_back
        print(f"4. 부모 함수 이름: {parent_frame.f_code.co_name}")
        print(f"5. 부모 함수의 지역 변수: {parent_frame.f_locals}")

    inner_func()

outer_func()
````
