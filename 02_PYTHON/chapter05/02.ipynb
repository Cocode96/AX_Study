#### 1. 전역 변수(Global)과 지역 변수(Local)
- 전역변수(Global) : 함수 외부에 정의된 변수, 코드 전역에서 접근 가능
- 지역변수(local) : 함수 내부에 정의된 변수, 함수가 실행될때만 유효한 변수

````python
a = 10 # 전역 변수

def func():
    b = 20 # 지역변수
    print(f"{a=}, {b=}")

# print(a, b)
print(a)
func()
````

````python
def outer():
    a = 10

    def inner():
        b = 20
        print(f"{a=}, {b=}")

    inner()

outer()
````

#### 2. 파이썬의 변수 검색 규칙 : LEGB 스코프

접근 우선순위 : 좁은 범위가 우선순위가 높다.
- L(Local - 지역변수) > E(Enclosing - 외부 함수 변수) > G(Global - 전역 변수) > B(Builtin - 내장 변수)

````python
a = 'Global'
def outer():
    a = 'Enclosing'
    print("outer : ", a)

    def inner():
        a = "local"
        print("inner : ", a)

    inner()

outer()
````

#### 3. global 키워드

````python
a = True
def outer():
    a = 'Enclosing'
    print("outer : ", a)

    def inner():
        global a # a라는 변수는 전역변수를 찾겠다
        a = not a
        print("inner : ", a)

    inner()

outer()
````

#### 4. 일급 함수(First-class Function)
- 함수와 변수를 동일하게 취급
- 함수의 매개변수로 함수를 사용할 수 있다. - 콜백 함수(다른 함수에 전달되어 나중에 불려오는 함수)
- 함수의 반환값으로 함수를 사용할 수 있다. - 팩토리 함수(객체 또는 함수를 만들어 반환하는 함수)

````python
def callback2():
    print("callback()")

def func(callback):
    print("func()")
    callback()

func(callback2)
````

````python
def outer(num1):
    def inner(num2):
        return num1 + num2

    return inner
````

````python
a = outer(10)
a(20)
````

````python
b = outer(20) # 항상 10을 더해주는 새로운 함수 해당 outer(20) 자체가 객체가 되서 힙영역에 할당된다고 보면된다. num1이 inner함수에 속박(closure)
b(20) # 10 + 20
````

````python
b(30)
````

````python
def change_numbers(*args, callback):
    numbers = list(args)

    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])

    return numbers

def square(num):
    return num ** 2

change_numbers(1, 2, 24, 44, 44, 4, callback=square)
````

````python
def plus(num):
    return num + 10

change_numbers(10, 20, 30, callback=plus)
````

#### 5. lambda함수
- 간단하게 일회용으로 사용하는 경우 한줄의 식으로 간단하게 함수 정의
- 일회용으로 사용되므로 매개변수 이름도 최대한 간결하게(한자)
```python
lambda 매개변수, 매개변수, ...: 처리 부분
```

````python
square = lambda num: num ** 2

square(10)
````

````python
change_numbers(10, 20, 30, callback=lambda num: num ** 2)
````

#### 7. lambda와 map(), filter()를 활용한 데이터 일괄 가공
- map() : 변환 작업
- filter() : 특정 조건에서 데이터를 선별

````python
fruits = ["사과", "오렌지", "망고"]
list(map(lambda x: f"**{x}**", fruits))
````

````python
numbers = [*range(1, 11)]
list(filter(lambda x : x % 2 == 1, numbers))
````
