#### 1. 이터러블(Iterable)
- 반복 가능한 객체
- \_\_iter\_\_(..) 정의되어 있다면 반복 가능!
- \_\_next\_\_(..) 인덱스를 한칸씩 이동하면서 요소를 꺼내오는 기능을 정의

```python
nums = [10, 20, 30]

nums.__iter__
```

```python
for num in nums:
    print(num)
```

#### 2. 이터레이터(Iterator)
- 이터레이터 객체 생성(__iter__ 호출(iter()))
- 순서를 한칸씩 이동하면서 요소 조회(__next__ 호출(next())

```python
fruits = ['apple', 'orange', 'Melon', 'Banana']

it = iter(fruits) # __iter__(..)호출
```

```python
next(it)
```

#### 3. 제너레이터(Generator)
- 함수의 실행을 정지하고 원할 때 다시 재개하는 형태

```python
def generate_nums():
    yield 10
    yield 20
    yield 30

generate_nums()
```

```python
gen = generate_nums()
gen
```

```python
next(gen) # 꺼내올때 마다 힙 영역에 해당 값을 할당
```

#### 4. 제네레이터 표현식
- 참고 : 리스트 컴프리헨션 -> 생성 결과가 처음부터 메모리에 적재
- 리스트 대신 튜플로 리스트 컴프리헨션과 동일하게 사용하면 된다. -> 메모리 소비가적다.
- 대용량 데이터 처리

```python
gen = (i for i in range(1, 101) if i % 2 == 0)
gen
```
