#### 1. 상속(Inheritance)
- 클래스 기반의 상속(상속에 대한 정의 -> 객체 생성(상속 관계 성립))
```python
class 자식클래스(부모클래스):
    ...
```
- super: 부모 클래스를 가르키는 참조 변수
- 참고) self : 자식 객체(현재 객체)의 참조 변수
- super() : 부모쪽 객체 생성

- 부모클래스 : 일반적인 정의들, 상위 개념
- 자식클래스 : 구체적인 정의들, 하위 개념

````python

class Animal:
    
    def __init__(self, kind):
        self.kind = kind # 동물 종류

    def move(self):
        print("동물이 움직인다.")

class Dog(Animal):

    def __init__(self):
        super().__init__("강아지")

class Cat(Animal):
    
    def __init__(self):
        super().__init__("고양이")
````

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
````

````python
print(dog.kind, cat.kind)
````

#### 2. 메서드 오버라이딩(Overriding)
- 메서드 재정의 
- 부모쪽 클래스와 동일한 이름으로 정의하여 기능을 수정, 재정의
- 자식쪽 클래스에 정의된 동일한 이름의 함수가 호출 우선순위를 가진다.

````python

class Animal:
    
    def __init__(self, kind):
        self.kind = kind # 동물 종류

    def move(self):
        print("동물이 움직인다.")

class Dog(Animal):

    def __init__(self):
        super().__init__("강아지")

    def move(self):
        super().move() # 부모쪽 객체의 move()
        print(self.kind + "가 걸음으로 움직인다.")
        


class Cat(Animal):
    
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print("고양이 걸음으로 움직인다.")
````

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
````

````python
class Human:
    def __init__(self):
        pass

    def __str__(self):
        pass

    def __repr__(self):
        pass

    def move(self):
        print("두발로 직립보행한다.")
````

````python
human = Human()
human.move()
````

#### 객체의 출처(클래스 자료형) 확인하는 함수 - isinstance(참조 변수, 클래스/자료형)

````python
print("Human은 Animal 클래스로 부터 생성되었나?", isinstance(human, Animal))
print("이것은 체크가 안되는 것이여", human is Animal)
print("이것은 체크가 안되는 것이여", dog is Dog)
print("dog는 Dog 클래스로 부터 생성되었나?", isinstance(dog, Dog))
print("dog는 Animal 클래스로 부터 생성되었나?", isinstance(dog, Animal))
print("cat은 Cat 클래스로 부터 생성되었나?", isinstance(cat, Cat))
print("cat은 Dog `클래스로 부터 생성되었나?", isinstance(cat, Dog))
````

````python
s = "ABC"
s.lower()
s.upper()

# 문자열인지 체크
print("s는 str 클래스로 부터 생성되었나?", isinstance(s, str))
````

#### 다중 상속 - 여러 부모클래스를 상속 관계로 정의 가능
- 다중 상속시 같은 메서드가 있을 때 호출 우선 순위(MRO)
- 자식 클래스의 메서드 -> 상속 관계에서 왼쪽에 있는 클래스 ->

````python
class ParentA:
    def a(self):
        print("ParentA - a()")
    
    def b(self):
        print("ParentA - b()")

class ParentB:
    def a(self):
        print("ParentB - a()")
    
    def c(self):
        print("ParentB - c()")

#class Child(ParentA, ParentB): # ParentA 의 우선순위가 높다 우선순위높음->낮음 순으로 다중 상속을 한다.
#    pass

class Child(ParentB, ParentA):

    def a(self):
        super().a()
        print("자식 객체에서 불려오는겁니다.")   
    pass

child = Child()
child.a()
child.b()
child.c()

Child.__mro__ # Method Resolution Order - 메서드 우선순위

````

#### 4. 정보 은닉과 캡슐화(Encapsulation)

````python
class Schedule:

    def  __init__(self, year, month, day):
        self._year = year
        self._month = month
        self._day = day

    def __str__(self):
        return f"year={self._year}, month={self._month}, day={self._day}"
    
    def __repr__(self):
        pass
````

````python
s1 = Schedule(2026, 8, 18)
print(s1)
````

````python
# 인스턴스 변수에 직접 대입이 가능하므로 값에 대한 통제는 불가!
# 통제 불가능하면 -> 직접 접근을 막아야 한다! -> 인스턴스 변수를 외부에서 접근하지 못하게 막는다.
s1._year = 2026
s1._month = 2
s1._day = 31

print(s1)
````

#### 5. 파이썬 프라이빗(private) 변수와 이름 변형 규칙(name mangling)
- \_\_변수명 : 클래스 내부에서만 접근 가능한 인스턴스 변수가 된다.

````python
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day

    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"

    def get_year(self):
        return self.__year
    
    def get_month(self):
        return self.__month
    
    def get_day(self):
        return self.__day

    def set_year(self, year):
        self.__year = year
    
    def set_month(self, month):
        self.__month = month
    
    def set_day(self, day):
        if self.__month == 2 and day > 28:
            day = 28
        self.__day = day
````

````python
s1 = Schedule(2026, 8, 18)
print(s1)

s1.__year = 2026
s1.__month = 2
s1.__day = 31

print(s1)

s1.set_year(2025)
s1.set_month(2)
s1.set_day(31)

print(s1)
````
