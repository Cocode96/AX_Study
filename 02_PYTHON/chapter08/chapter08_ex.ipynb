```python
class Book:
    def __init__(self, title, author):
        self._title = title
        self._author = author
    
    def info(self):
        return print(f"책 제목 : [{self._title}], 저자 : [{self._author}]")

book1 = Book("전문가를 위한 C++", "C++고수")
book2 = Book("전문가를 위한 파이썬", "파이썬 고수")

book1.info()
book2.info()
```

```python
class SmartLamp:
    def __init__(self, model, is_on=False):
        self._model = model
        self._is_on = is_on
    
    def turn_on(self):
        self._is_on = True
        print(f"[{self._model}] 조명이 켜졌습니다.")
    
    def turn_off(self):
        self._is_on = False
        print(f"[{self._model}] 조명이 꺼졌습니다.")

lamp = SmartLamp("최신 모델 사무실용 전등")
lamp.turn_on()
lamp.turn_off()
```

```python
class Hero:
    def __init__(self, name, hp=100):
        self._name = name
        self._hp = hp
    
    def take_damage(self, damage):
        self._hp -= damage
        print(f"Hp 잔량 : {self._hp}")

        if self._hp < 0:
            print(f"{self._name} 캐릭터가 쓰러졌습니다.")

mag = Hero("메이지")
for i in range(1, 7):
    mag.take_damage(i * 5)
```

```python
class Phone:
    def call(self):
        print("통화를 시작합니다.")

class SmartPhone(Phone):
    def call(self):
        print("영상 통화를 시작합니다.")
    
    def search_web(self):
        print("인터넷 검색을 시작합니다.")
    
galaxy_smart_phone = SmartPhone()
galaxy_smart_phone.call()
galaxy_smart_phone.search_web()
```

```python
class DoorLock:
    def __init__(self, password):
        self.__password = password
    
    def get_password(self):
        return self.__password
    
    def set_password(self, password):
        if len(password) >= 4:
            self.__password = password
            print("정상 변경되었습니다.")
            return
        else:
            print("경고: 비밀번호는 4자리 이상이어야 합니다.")
            return

dl = DoorLock("1234")
dl.set_password("9876")
print(dl.get_password())
```
