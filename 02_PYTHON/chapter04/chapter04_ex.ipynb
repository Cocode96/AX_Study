```python
current_temp = 32

if current_temp >= 30:
    print("폭염 주의보가 발령되었습니다.")
else:
    print("쾌적한 날씨입니다.")
```

```python
weight = 23

if weight > 23:
    print("대형 수하물 대상입니다. 추가요금이 부과됩니다.")
elif weight > 15:
    print("일반 위탁 수하물 대상입니다.")
else:
    print("무료 수하물 대상입니다.")
```

```python
age = 14
height = 135

if age >= 12 and height >= 130:
    print("탑승 가능합니다. 관리자 확인을 진행하십시오.")
else:
    print("탑승 요건을 만족하지 못했습니다.")
```
