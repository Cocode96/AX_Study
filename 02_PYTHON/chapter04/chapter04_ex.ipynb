```python
order_database = [
    {"name": "김철수", "completed": True, "price": 60000},
    {"name": "이영희", "completed": False, "price": 35000},
    {"name": "박민수", "completed": True, "price": 25000},
    {"name": "최지영", "completed": True, "price": 85000},
    {"name": "정성민", "completed": False, "price": 90000}
]

total_price = 0
order_cnt = 0

for order in order_database:
    name, completed, price = order.values()
    if completed == True and price >= 50000:
        total_price += price
        order_cnt += 1
    
print(f"우수 고객 주문 건수 : {order_cnt}\n우수 고객총 결제 금액 : {total_price}")
```

```python
fruit_sweetness = [12.5, 14.2, 11.8, 10.5, 15.0, 13.6, 11.2, 14.8, 12.0, 13.1]

highest_fruit_sweetness = fruit_sweetness[0]
lowest_fruit_sweetness = fruit_sweetness[0]
total_furit_sweetness = 0

for sweetness in fruit_sweetness:
    if sweetness > highest_fruit_sweetness:
        highest_fruit_sweetness = sweetness
    if sweetness < lowest_fruit_sweetness:
        lowest_fruit_sweetness = sweetness

    total_furit_sweetness += sweetness

avg_furit_sweetness = total_furit_sweetness / len(fruit_sweetness)

print(f"최고 과일 당도: {highest_fruit_sweetness}")
print(f"최저 과일 당도: {lowest_fruit_sweetness}")
print(f"과일 평균 당도 : {avg_furit_sweetness}")
```

```python
import random

answer = random.randint(1, 50)
try_cnt = 0

print("보물 찾기 게임")
print("보물은 1번부터 50번방 중에 있습니다.")
print("보물을 찾아주세요")

while True:
    print("보물을 찾을 방 번호 (종료 : q) : ", end="")
    player_input = input()

    if player_input == "q":
        break

    number = int(player_input)

    if number == answer:
        print("정답입니다!")
        break
    elif number < answer:
        print(f"보물은 {player_input}보다 더 큰 숫자의 방에 있습니다")
    elif number > answer:
        print(f"보물은 {player_input}보다 더 작은 숫자의 방에 있습니다.")

    try_cnt += 1

print(f"축하합니다! {try_cnt}회 만에 보물을 찾았습니다.")
```
