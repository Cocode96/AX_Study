#### 1. for 문

#### 횟수가 정해져 있는 반복
'''
for num in range(반복횟수):
    # 반복 수행되는 코드
'''

- i : 순서를 의미하는 index, 관례, 다음 for문은 알파벳 순서대로 j,k,l,..
- range(반복횟수) : 0 ~ 반복 횟수
- range(시작, 종료) : 시작 이상 종료 미만 range(2, 5) -> 2, 3, 4
- range(시작, 종료, 증감단위)


```python
# 참고 - range() 숫자 생성시에도 사용
nums = [*range(1, 101)]

nums[:5]
```

```python
total = 0
start = 1

for i in range(100):
    #total = total + start
    total += start
    start += 1
    print("반복횟수 : ", i)

print(f"1~100 합계 : {total}")
```

```python
# 반복 횟수를 사용하지 않는 경우
for _ in range(10):
    print("반복")
```

```python
fruits = [
    '사과',
    '포도',
    '수박',
    '배',
    '오렌지',
    '망고',
    '멜론'
]
fruits_count = 0

print(*fruits)


for i in range(0, 7):
    fruits_count += 1
    print(fruits[i])

print(f"과일의 개수 : {fruits_count}")
```

```python
fruits = [
    '사과',
    '포도',
    '수박',
    '배',
    '오렌지',
    '망고',
    '멜론'
]

for i in range(0, 7, 2): # 0, 2, 4, 6
    print(fruits)
```
