# 4장 1강 : 조건문
----
- 판별식에 따라서 참(True)면 참에 실행되는 코드를 정의하고, 거짓(False)이면 거짓 조건에 실행되는 코드를 정의할 수 있는 문장!
- 판별식 : 비교 연산자, 논리 연산자, 값

#### 1. 단일 조건문 : if
if 판별식:
    판별식이 참일때 실행되는 코드

파이썬은 c++처럼 중괄호로 지역을 나누는게 아니라 들여쓰기로 하위 상위를 구분한다.

```python
num = 10
if num > 5:
    print("5보다 크다.")
```

#### 2. if-else 조건문

```python
num = 10
if num < 10:
    print(f"{num}은 10보다 작다")
else:
    print(f"{num}은 10이상이다.")
```

#### 3. 다중 조건 분기 : if-elif-else

```python
age = int(input())
if age >= 20:
    print("성인")
elif age >= 17:
    print("고등학생")
elif age >= 14:
    print("중학생")
elif age >= 8:
    print("초등학생")
else: # 앞선 모든 조건문이 False 이면 실행되는 구문
    print("유아")
```

#### 4. 복잡한 조건 처리 : 중첩 조건문과 논리 연산자

```python
#입력 받는 숫자가 10이상 100 미만인지 확인
num = int(input())
if num >= 10:
    if num < 100:
        print(f"{num}는 10이상 100미만 입니다.")
```
