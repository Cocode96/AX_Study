```python
import csv

frutis_info = [["사과", "사과는 달아"], ["배", "배는 시원해"], ["오렌지", "오렌지는 셔"]]

with open("fruit_inventory.csv", "w", encoding="UTF-8", newline="") as fp:
    writer = csv.writer(fp)
    writer.writerows(frutis_info)

with open("fruit_inventory.csv", "r", encoding="UTF-8", newline="") as fp:
    reader = csv.reader(fp)
    print(*list(reader))
```

```python
import json

player_info = {"캐릭터 이름" : "석호", "레벨" : 10, "관리자 권한" : True}

with open("character_profile.json", "w", encoding="UTF-8") as fp:
    json.dump(player_info, fp, indent=4, ensure_ascii=False)

with open("character_profile.json", "r", encoding="UTF-8") as fp:
    player_info_json = json.load(fp)
    print(player_info_json, type(player_info_json))
```
