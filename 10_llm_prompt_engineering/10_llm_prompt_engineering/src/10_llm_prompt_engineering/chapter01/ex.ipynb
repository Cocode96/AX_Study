```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser

# 1. 답변 규칙과 질문이 들어갈 자리를 만듭니다.
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "한국어로 객관적인 사실만 전달하세요. "
        "모르는 내용은 '확인할 수 없습니다'라고 명확히 답변하세요. "
        "질문에 잘못된 전제가 있으면 이를 지적하고, "
        "존재하지 않는 사실을 만들어내지 마세요."
        "문장마다 줄바꿈하여 답하세요."
    ),
    ("human", "{question}")
])

# 2. Ollama의 Mistral 모델을 연결합니다.
model = ChatOllama(
    model="mistral",
    temperature=0
)

# 3. 응답에서 문자열을 추출하고 체인으로 연결합니다.
parser = StrOutputParser()
chain = prompt | model | parser

# 4. 질문을 전달하여 실행합니다.
result = chain.invoke({
    "question": "1999년 조선시대에 발명된 스마트폰의 기능에 대해 설명해줘"
})

print(result)
```

```python
import re

def mask_personal_information(text: str) -> str:
    # 0000-0000-0000-0000 형태의 16자리 카드번호 감자하는 패턴입니다.
    card_pattern = re.compile(r'\d{4}-\d{4}-\d{4}-\d{4}')

    # 일치하는 패턴을 대체 문구로 치환합니다.
    masked = card_pattern.sub("[카드번호 보호]", text)
    return masked


mask_personal_information("결제 오류가 발생했습니다 제 카드 번호는 1234-5678-9012-3456 입니다 확인 부탁드립니다")
```
